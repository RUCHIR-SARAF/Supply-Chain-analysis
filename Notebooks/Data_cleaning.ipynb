### Business Context and Objectives
the company operates a global ecom platform, selling products across categories including sporting goods fitness equip, outdoor gear, footwear, apparel across multiple international region.
### Core business problem -
actual shipping time deviates from the delivery windows creating eroded customer trust, reduced order profitability to make reliable commitments to buyers at point of purchase

### Analytical Objectives-
- Understand current state if delivery performance across all dimensions
- Quantify the financial impact of delays on order profitability
- Identify the primary operational bottlenecks driving late deliveries]
- Build a predictive model to flag high risk orders before they are shipped
- Deliver actionable recommendations to improve on-time delivery and profitability
